**Homodyne**

For the homodyne detection we measure the quadratures, X in this case.

For the homodyne detection we consider two states alpha and -alpha. This states are overlaped and it's distributions too.

This means that when we sometimes measure alpha, we could actually be measuring -$\alpha$ and this creates and error. This error is given by the sum of $$P(B)*P(A|B)+P(A)*P(B|A)$$

This error value can also correspond to the are of the graphis where they are overlaped, since this area represents the probability of that.

The probability of measuring the X quadrature is given by $$P_A(X) = |\langle X | \alpha \rangle|^2 = \left| \psi_\alpha(X) \right|^2 = \left| \sum_{n=0}^{\infty} \langle X | n \rangle \langle n | \alpha \rangle \right|^2$$

Now, let's plot the probability-function in function of X, to confirm it's a gaussian.

And now let's try to calculate the error probability for two random states and also draw the error graphic.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
alpha_plus = 1    #Bit0, vaccum state
alpha_minus = -1    #Bit1, the more far away this values are, the lower the sobrepostition
                #value is and so the lower the error

mu0 = np.sqrt(2) * np.real(alpha_plus)  #average value of the vacuum state
mu1 = np.sqrt(2) * np.real(alpha_minus) #aberage value of minus alpha

sigma = 0.5  # standard deviation for the quadratures of the alpha states

P0 = norm.pdf(X, mu0, sigma)  #Normal distribution
P1 = norm.pdf(X, mu1, sigma)
```

```python
#Plotting
plt.figure(figsize=(9, 5))
plt.plot(X, P0, label=r'$P(X|-\alpha)$', color='blue', linewidth=2)
plt.plot(X, P1, label=r'$P(X|\alpha)$', color='orange', linewidth=2)
plt.title('Distributions of the probabilities')
plt.xlabel('Quadrature $X$')
plt.ylabel('Density probabilitie')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

And the probability of this error is given by the area of the overlaped areas, aproximatelly between x=0 and x=2.3.

```python
from scipy.integrate import simpson

interception = np.minimum(P0, P1) #Finding the minimum value where they intersept

area_total = simpson(y=interception, x=X) #Calculating the total area of interseption

P_error = 0.5 * area_total #dividing by 2 since the error is 50/50 for both bits

print(f"Total Area: {area_total:.5f}")
print(f"Error probability (P_e): {P_error:.5f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erfc
```

```python
# Range of coherent state amplitudes
alphas = np.linspace(0, 2, 200)

# Kennedy receiver error probability
P_kennedy = 0.5 * np.exp(-4 * np.abs(alphas)**2)

# Plot
plt.figure(figsize=(8, 5))

plt.plot(alphas, P_kennedy, label="Kennedy receiver")

plt.xlabel(r"Coherent state amplitude $|\alpha|$")
plt.ylabel(r"Error probability $P_{\mathrm{Kennedy}}$")
plt.title("Kennedy Receiver Error Probability")

plt.grid(True)
plt.legend()
plt.show()
```

## Plot P_err comparison for Homodyne and Kennedy

```python
# Range of coherent state amplitudes
alphas = np.linspace(0, 2, 200)

# Homodyne receiver error probability
P_hom_theory = 0.5 * erfc(np.sqrt(2) * np.abs(alphas))

# Kennedy receiver error probability
P_kennedy = 0.5 * np.exp(-4 * np.abs(alphas)**2)

# Plot
plt.figure(figsize=(8, 5))

plt.plot(alphas, P_hom_theory, label="Homodyne receiver")

plt.plot(alphas, P_kennedy, label="Kennedy receiver")

plt.xlabel(r"Coherent state amplitude $|\alpha|$")
plt.ylabel("Error probability")
plt.title("Homodyne and Kennedy Receiver Performance")

plt.grid(True)
plt.legend()
plt.show()
```

```python
# Helstrom bound
overlap_squared = np.exp(-4 * np.abs(alphas)**2)

P_helstrom = 0.5 * (1 - np.sqrt(1 - overlap_squared))

plt.figure(figsize=(8, 5))
plt.plot(alphas, P_hom_theory, label="Homodyne receiver")
plt.plot(alphas,P_kennedy, label="Kennedy receiver")
plt.plot(alphas, P_helstrom, linestyle="--", label="Helstrom bound")

plt.xlabel(r"Coherent state amplitude $|\alpha|$")
plt.ylabel("Error probability")
plt.title("Binary Coherent-State Discrimination")

plt.grid(True)
plt.legend()
plt.show()
```
